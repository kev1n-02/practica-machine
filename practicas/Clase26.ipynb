```python
#importamos las librerias
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Creamos el conjunto de datos
df = pd.DataFrame({
    "horas":[1,2,3,4,5],
    "nota":[2,3,5,4,6]
})
x= df[["horas"]]
y= df[["nota"]]
print (x)
```

```python
#creamos nuestro modelo
modelo = LinearRegression()
modelo.fit(x, y)
```

```python
#imprimimos el interceptor y el coef
print("interceptor: ", modelo.intercept_)
print("Pendiente: ", modelo.coef_[0]) 
```

```python
#Valores predichos
y_pred = modelo.predict(x)
```

```python
#Realizamos la nueva prediccion 7
horas=float(input("Ingrese las horas de estudio"))
prediccion = modelo.predict([[horas]])
print("Prediccion:" , prediccion)
```

```python
# Convertimos la predicción en un número
nota_predicha = prediccion[0][0]

# Valores para dibujar la línea de regresión
x_linea = np.linspace(1, 6, 100)

# Creamos DataFrame para evitar el warning
x_linea_df = pd.DataFrame({"horas": x_linea})

y_linea = modelo.predict(x_linea_df)

plt.figure(figsize=(8, 5))

# Valores reales de la tabla
plt.scatter(
    df["horas"],
    df["nota"],
    color="yellow",
    s=80,
    label="Valores reales"
)

# Predicciones de los datos de la tabla
plt.scatter(
    df["horas"],
    y_pred,
    color="green",
    s=80,
    label="Predicciones de la tabla"
)

# Nueva predicción: 6 horas
plt.scatter(
    horas,
    nota_predicha,
    color="orange",
    s=120,
    label="Nueva predicción"
)

# Línea de regresión
plt.plot(
    x_linea,
    y_linea,
    color="blue",
    linewidth=2,
    label="Regresión lineal"
)

# Texto de la nueva predicción
plt.annotate(
    f"({horas}, {nota_predicha:.2f})",
    (horas, nota_predicha),
    xytext=(10, 10),
    textcoords="offset points"
)

plt.xlabel("Horas de estudio")
plt.ylabel("Nota")
plt.title("Regresión Lineal: Horas de estudio vs Nota")

plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```
