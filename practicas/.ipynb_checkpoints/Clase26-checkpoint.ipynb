```python
#importamos las librerias
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Creamos el conjunto de datos
df = pd.DataFrame({
    "horas":[1,2,3,4,5],
    "nota":[2,3,5,4,6]
})
x= df[["horas"]]
y= df[["nota"]]
print (x)
```

```python
#creamos nuestro modelo
modelo = LinearRegression()
modelo.fit(x, y)
```

```python
#imprimimos el interceptor y el coef
print("interceptor: ", modelo.intercept_)
print("Pendiente: ", modelo.coef_[0]) 
```

```python
#Valores predichos
y_pred = modelo.predict(x)
```

```python
#Realizamos la nueva prediccion 7
horas=float(input("Ingrese las horas de estudio"))
prediccion = modelo.predict([[horas]])
print("Prediccion:" , prediccion)
```

```python
# Convertimos la predicción en un número
nota_predicha = prediccion[0][0]

# Valores para dibujar la línea de regresión
x_linea = np.linspace(df["horas"].min(), horas, 100)

# Creamos DataFrame para hacer la predicción
x_linea_df = pd.DataFrame({"horas": x_linea})

# Calculamos los valores de la línea
y_linea = modelo.predict(x_linea_df)

# Creamos el gráfico
plt.figure(figsize=(9, 6))

# Datos reales
plt.scatter(
    df["horas"],
    df["nota"],
    s=80,
    label="Datos reales"
)

# Línea de regresión
plt.plot(
    x_linea,
    y_linea,
    linewidth=2,
    label="Línea de regresión"
)

# Nueva predicción
plt.scatter(
    horas,
    nota_predicha,
    s=150,
    marker="X",
    label=f"Predicción ({horas} horas)"
)

# Línea vertical de referencia
plt.axvline(
    x=horas,
    linestyle="--",
    alpha=0.5
)

# Línea horizontal de referencia
plt.axhline(
    y=nota_predicha,
    linestyle="--",
    alpha=0.5
)

# Texto de la predicción
plt.annotate(
    f"({horas}, {nota_predicha:.2f})",
    xy=(horas, nota_predicha),
    xytext=(-80, 20),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

# Títulos y etiquetas
plt.xlabel("Horas de estudio")
plt.ylabel("Nota")
plt.title("Regresión Lineal: Horas de estudio vs. Nota")

plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```
