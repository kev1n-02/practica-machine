#### Strings

-  Cadenas de caracteres.
-  Son secuencias: la posición de los caracteres es importante.
-  Son immutables: las operaciones sobre strings no cambian el string original.

```python
s = 'John "ee" Doe'
print(s[0])     # Primer carácter del string.
print(s[-1])    # Último carácter del string.
print(s[1:8:2]) # Substring desde el segundo carácter (inclusive) hasta el octavo (exclusive). Esta técnica se la conoce como 'slicing'.
print(s[:])     # Todo el string.
print(s + "e")  # Concatenación.
```

#### 2.4 Conversión entre tipos
-  Aveces queremos que un objeto sea de un tipo específico.
-  Podemos obtener objetos de un tipo a partir de objetos de un tipo diferente (casting).

```python
a = int(2.8)   # a será 2
b = int("3")   # b será 3
c = float(1)   # c será 1.0
d = float("3") # d será 3.0
e = str(2)     # e será '2'
f = str(3.0)   # f será '3.0'
g = bool("a")  # g será True
h = bool("")   # h será False
i = bool(3)    # i será True
j = bool(0)    # j será False
k = bool(None)

print(a)
print(type(a))
print(b)
print(type(b))
print(c)
print(type(c))
print(d)
print(type(d))
print(e)
print(type(e))
print(f)
print(type(f))
print(g)
print(type(g))
print(h)
print(type(h))
print(i)
print(type(i))
print(j)
print(type(j))
print(k)
```

```python
print(7/4) # Divisiónconvencional.Resultadodetipo'float'
print(7//4) # Divisiónentera.Resultadodetipo'int'
print(int(7/4)) # Divisónconvencional.Conversióndelresultadode'float'a'int'
```

#### 2.5 Operadores
[Python 3 precedencia en operaciones](https://docs.python.org/3/reference/expressions.html#operator-precedence)
- Combinación de valores, variables y operadores.
- Operadores y operandos.
#### Operadores aritméticos
| Operador | Descripción |
| :---: | :--- |
| `a + b` | Suma |
| `a - b` | Resta |
| `a / b` | División |
| `a // b` | División Entera |
| `a % b` | Módulo / Resto |
| `a * b` | Multiplicación |
| `a ** b` | Exponenciación |

```python
x =3
y =2

print('x + y = ', x + y)
print('x- y = ', x- y)
print('x * y = ', x * y)
print('x / y = ', x / y)
print('x // y = ', x // y)
print('x % y = ', x % y)
print('x ** y = ', x ** y)
```

#### Operadores de comparación 
| Operador | Descripción |
| :---: | :--- |
| `a > b` | Mayor |
| `a < b` | Menor |
| `a == b` | Igualdad |
| `a != b` | Desigualdad |
| `a >= b` | Mayor o Igual |
| `a <= b` | Menor o Igual |

```python
x = 10
y = 12

print('x > y es ', x > y)
print('x < y es ', x < y)
print('x == y es ', x == y)
print('x != y es ', x != y)
print('x >= y es ', x >= y)
print('x <= y es ', x <= y)
```

#### Operadores Lógicos

| Operador | Descripción |
| :---: | :--- |
| `a and b` | True, si ambos son True |
| `a or b` | True, si alguno de los dos es True |
| `a ^ b` | XOR - True, si solo uno de los dos es True |
| `not a` | Negación |

Enlace a [Tablas de Verdad](https://en.wikipedia.org/wiki/Truth_table)

```python
x = True
y = False

print('x and y es :', x and y)
print('x or y es :', x or y)
print('x xor y es :', x ^ y)
print('not x es :', not x)
```

#### Operadore de Pertenencia
| Operador | Descripción |
| :---: | :--- |
| `a in b` | True, si a se encuentra en la secuencia b |
| `a not in b` | True, si a no se encuentra en la secuencia b |

```python
x = 'Hola Mundo'
y = {1:'a',2:'b'}

print('H' in x)        # True
print('hola' not in x) # True

print(1 in y)          # True
print('a' in y)        # False
```

#### 3 Tipos de datos compuestos (colecciones)
#### 3.1 Listas
-  Una colección de objetos.
-  Mutables.
-  Tipos arbitrarios heterogeneos.
-  Puede contener duplicados.
-  Notienen tamaño fijo. Pueden contener tantos elementos como quepan en memoria.
-  Los elementos van ordenados por posición.
-  Se acceden usando la sintaxis: [index].
-  Los índices van de 0 a n-1, donde n es el número de elementos de la lista.
-  Son un tipo de Secuencia, al igual que los strings; por lo tanto, el orden (es decir, la posición de los objetos de la lista) es importante.
-  Soportan anidamiento.
-  Son una implementación del tipo abstracto de datos: Array Dinámico.

#### Diccionarios
- Colección de parejas clave-valor.
- Son mutables.
- Claves:
     - Cualquier objeto inmutable.
     - Sólo pueden aparecer una vez en el diccionario.
- Valores:
     - Sin restricciones. Cualquier objeto (enteros, strings, listas, etc.) puede hacer de valor.
-  Desde la versión 3.7 están ordenados.
-  Se pueden ver como listas indexadas por cualquier objeto inmutable, no necesariamente por números enteros.
-  A diferencia de las listas, no son secuencias. Son mappings.

#### Operaciones con diccionarios
- Creación de diccionarios.

```python
# Creación simple, usando una expresión literal.
persona = {'DNI' : '11111111D', 'Nombre' : 'Carlos', 'Edad' : 34}
print(persona)
```

```python
# Creación uniendo dos colecciones.

nombres = ['Pablo', 'Manolo', 'Pepe', 'Juan']
edades = [52, 14, 65]
24
datos = dict(zip(nombres,edades))
print(datos)
```

```python
# Creación pasando claves y valores a la función 'dict'
persona2 = dict(nombre='Rosa', apellido='Garcia')
print(persona2)
```

```python
# Creación usando una lista de tuplas de dos elementos.
persona2 = dict([('nombre', 'Rosa'), ('apellido','Garcia')])
print(persona2)
```

```python
# Creación incremental por medio de asignación (como las claves no existen, se crean nuevos items)
persona = {}
print(persona['DNI'])
persona['DNI'] = '11111111D'
persona['Nombre'] = 'Carlos'
persona['Edad'] = 34
persona['DNI'] = '222222222'
print(persona)
```

#### 3.3 Sets (Conjuntos)
Al igual que las listas:
- Colección de elementos.
- Tipos arbitrarios.
- Mutables.
- Notienen tamaño fijo. Pueden contener tantos elementos como quepan en memoria.

A diferencia de las listas:
- No puede tener duplicados.
- Se definen por medio de llaves.
- Los elementos no van ordenados por posición. No hay orden establecido.
- Solo pueden contener objetos inmutables.
- No soportan anidamiento.

#### 3.4 Tuplas
Al igual que las listas:
- Colección de elementos.
- Tipos arbitrarios.
- Puede contener duplicados.
- No tienen tamaño fijo. Pueden contener tantos elementos como quepan en memoria.
- Los elementos van ordenados por posición.
- Acceso a través de la sintaxis: [index]
- Índices van de 0 a n-1, donde n es el número de elementos de la tupla.
- Son secuencias donde el orden de los elementos importa.
- Soportan anidamiento.
  
A diferencia de las listas:
- Se definen por medio de paréntesis.
- Inmutables.

#### ¿Por qué tuplas?
- Representación de una colección fija de elementos (por ejemplo, una fecha).
- Pueden usarse en contextos que requieren inmutabilidad (por ejemplo, como claves de un
diccionario).

#### 4 Soluciones 

```python
# Se declara una variable entera.
x = 3

# Se declara una variable de tipo cadena.
y = "abc"

# Se convierte el número a cadena para poder concatenarlo con el texto
# y luego se muestra el resultado.
print(str(x) + y)
```

```python
# Ejercicio 2
# Se definen el número a evaluar y los límites del intervalo.
N = 5
a = 0
b = 6

# Se verifica que N sea mayor o igual al límite inferior
# y menor que el límite superior.
print(a <= N and N < b)
```

```python
# Ejercicio 3
# Se definen las cadenas de texto.
S1 = "Java"
S2 = "JavaScript"

# Se establecen las posiciones inicial y final del substring.
N1 = 0
N2 = 3

# Se extrae el substring de S1 y se verifica si está contenido en S2.
print(S1[N1:N2 + 1] in S2)
```

```python
# Ejercicio 4
# Lista con elementos repetidos.
L = [1, 2, 2, 3, 3, 2, 3, 1, 4, 5, 4, 5, 5]

# Se convierte la lista en un conjunto para eliminar los duplicados
# y posteriormente se vuelve a convertir en lista.
print(list(set(L)))
```

```python
# Ejercicio 5
# Reemplazar un elemento de una lista por otro.

# Lista original.
L = ["Python", "Java", "C++", "Kotlin"]

# Elemento que será reemplazado.
s = "Java"

# Nuevo elemento.
t = "JavaScript"

# Se obtiene la posición del elemento a reemplazar.
indice = L.index(s)

# Se reemplaza el elemento por el nuevo valor.
L[indice] = t

# Se muestra la lista actualizada.
print(L)
```

```python
# Ejercicio 6
# Reemplazar el último elemento de una tupla.

# Se define la tupla original.
tupla = (1, 2, 3, 4)

# Como las tuplas son inmutables, se convierten en lista.
lista = list(tupla)

# Se modifica el último elemento.
lista[-1] = 5

# Se convierte nuevamente la lista en tupla y se imprime.
print(tuple(lista))
```

```python
# Ejercicio 7
# Obtener la lista [8, 6, 4, 2] usando slicing.

# Lista original.
lista = [1, 2, 3, 4, 5, 6, 7, 8]

# Se recorre la lista desde el último elemento
# avanzando de dos en dos hacia atrás.
print(lista[-1::-2])
```

```python
# Ejercicio 8
# Eliminar un elemento de una tupla usando únicamente tuplas.

# Tupla original.
tupla = ("a", "b", "c", "d", "e", "f")

# Índice del elemento que se eliminará.
i = 3

# Se unen los elementos antes y después del índice indicado,
# formando una nueva tupla sin el elemento eliminado.
print(tupla[:i] + tupla[i + 1:])
```

```python
# Ejercicio 9
# Obtener la mediana de una lista de números.

# Lista de números.
L = [21, 9, 1, 5.5, 19, 3, 15]

# Se ordena la lista de menor a mayor.
L.sort()

# Se muestra la lista ordenada.
print(L)

# Se obtiene el elemento central de la lista,
# correspondiente a la mediana.
mediana = L[len(L) // 2]

# Se imprime la mediana.
print(mediana)
```
